```python
import torch
import numpy as np
import pandas as pd

from pymatgen.core import Structure
from pymatgen.io.cif import CifWriter

def generate_cif(row, supercell_size = [2,2,2]):
    """Generate cif file from row of dataframe."""
    crystal_str = row['cif']
    structure = Structure.from_str(crystal_str, fmt='cif')
    structure *= supercell_size
    cif_writer = CifWriter(structure)
    cif_content = str(cif_writer)
    return cif_content
```

```python
######################################################################################
# Generate supercell perov_5.
######################################################################################

for i in np.arange(2,6):

    supercell_size = [i,i,i]

    # test_zpc.
    df = pd.read_csv('./test_zpc.csv')  
    df['cif'] = df.apply(generate_cif, axis=1, supercell_size=supercell_size)
    output_csv_file = f'./cs{i}_test_zpc.csv'
    df.to_csv(output_csv_file, index=False)
    print(f"CSV with updated CIF ({i} {i} {i}) saved to {output_csv_file}")

    # train.
    df = pd.read_csv('./train.csv')
    df['cif'] = df.apply(generate_cif, axis=1, supercell_size=supercell_size)
    output_csv_file = f'./cs{i}_train.csv'
    df.to_csv(output_csv_file, index=False)
    print(f"CSV with updated CIF ({i} {i} {i}) saved to {output_csv_file}")
    
    # val.
    df = pd.read_csv('./val.csv')
    df['cif'] = df.apply(generate_cif, axis=1, supercell_size=supercell_size)
    output_csv_file = f'./cs{i}_val.csv'
    df.to_csv(output_csv_file, index=False)
    print(f"CSV with updated CIF ({i} {i} {i}) saved to {output_csv_file}")
```

```python
# Concatenate all generated supercell data accordingly.
df = pd.read_csv('./test_zpc.csv')
df_cs2 = pd.read_csv('./cs2_test_zpc.csv')
df_cs3 = pd.read_csv('./cs3_test_zpc.csv')
df_cs4 = pd.read_csv('./cs4_test_zpc.csv')
df_cs5 = pd.read_csv('./cs5_test_zpc.csv')
df_cat = pd.concat([df, df_cs2, df_cs3, df_cs4, df_cs5])

# Save.
output_csv_file = f'./cs_cat_test_zpc.csv'
df_cat.to_csv(output_csv_file, index=False)

# Screen.
print(len(df), len(df_cs2), len(df_cs3), len(df_cs4), len(df_cs5), len(df_cat))
```
