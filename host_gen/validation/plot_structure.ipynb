```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os.path as osp

from pymatgen.core import Structure
from pymatgen.io.ase import AseAtomsAdaptor

from ase.visualize.plot import plot_atoms
```

```python
file_names = [
    'data_recon_no_scaling_step_1.csv',
    'data_recon_with_scaling_step_1.csv',
    'data_recon_no_scaling_step_5.csv',
    'data_recon_with_scaling_step_5.csv',
    # 'data_recon_no_scaling_step_10.csv',
]
```

```python
adaptor = AseAtomsAdaptor()

plt.subplots(4,10,figsize=(20,8))
for name in file_names:
    df = pd.read_csv(osp.join('../data/recon', name), keep_default_na=False, na_values=[""])
    s_list = [Structure.from_str(s, fmt='cif', frac_tolerance=0, site_tolerance=0) for s in df['cif_recon'].to_list()]
    s_list = s_list[:10]

    for s in s_list:
        ax = plt.subplot(4,10,file_names.index(name)*10+s_list.index(s)+1)
        atoms = adaptor.get_atoms(s)
        plot_atoms(atoms, ax, radii=0.5, rotation=('-0x,0y,00z'))
        ax.axis('off')
```
