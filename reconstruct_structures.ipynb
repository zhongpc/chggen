```python
from pathlib import Path
from datetime import datetime
from typing import List
from types import SimpleNamespace
import pickle
import numpy as np
import torch

from pymatgen.core import Structure, Lattice, PeriodicSite, Element
```

```python

device = torch.device("cuda:7")

with open('./output_from_ld', 'rb') as fp:
    output = pickle.load(fp)

```

```python
frac_coords = output['frac_coords'].to('cpu').numpy()

print(frac_coords.shape)

num_atoms = output['num_atoms'].to('cpu').numpy().reshape(-1)

atom_types = output['atom_types'].to('cpu').numpy().reshape(-1)
lengths = output['lengths'].to('cpu').numpy()
print(lengths.shape)
angles = output['angles'].to('cpu').numpy()


print(len(atom_types))
```

```python
type_index = 0


for ii in range(0, len(lengths)):
    print(ii)
    
    num_ = num_atoms[ii]
    atom_type = atom_types[type_index:(type_index+num_)]
    
    FracCoords =  frac_coords[type_index:(type_index+num_), :]
    print(FracCoords.shape)
    Species = [Element.from_Z(ele_Z) for ele_Z in atom_type]; 
    print(len(Species))
    Latt = Lattice.from_parameters(a = lengths[ii,0], b = lengths[ii,1], c = lengths[ii,2],
                                   alpha= angles[ii, 0], beta= angles[ii,1], gamma=angles[ii, 2])
    
    s_gen = Structure(lattice= Latt , species= Species, coords= FracCoords,
                      to_unit_cell=False,coords_are_cartesian=False);
    print(s_gen.composition)
    s_gen.to(filename= './gen_structures/test_' + str(ii) + '.cif')
    
    type_index += num_
    
```
